```python
import pandas as pd
import numpy as np

data = {
    'Employee_ID': [101, 102, 103, 104, 105, 106, 107, 108],
    # Numerical columns (with NaN)
    'Age': [25.0, 30.0, np.nan, 35.0, 40.0, np.nan, 28.0, 32.0],
    'Salary': [50000.0, np.nan, 65000.0, 80000.0, np.nan, 72000.0, 58000.0, 62000.0],
    'Experience_Years': [2.0, 5.0, 7.0, np.nan, 12.0, 8.0, np.nan, 6.0],
    
    # Categorical columns (with None / NaN)
    'Department': ['IT', 'HR', np.nan, 'Finance', 'IT', 'Finance', np.nan, 'HR'],
    'City': ['Mumbai', np.nan, 'Delhi', 'Bangalore', np.nan, 'Mumbai', 'Pune', 'Delhi'],
    'Education': ['Bachelors', 'Masters', 'Masters', np.nan, 'PhD', 'Bachelors', np.nan, 'Masters']
}

df = pd.DataFrame(data)
df
```

```python
from sklearn.impute import SimpleImputer
```

```python

imputer=SimpleImputer(strategy='mean')

imputer.fit_transform(df[['Salary']])
```

```python
imputer=SimpleImputer(strategy='median')
df['new_Experience_Years']=imputer.fit_transform(df[['Experience_Years']]).flatten()
```

```python
df
```

```python
imputer=SimpleImputer(strategy='most_frequent',)
df['City']=imputer.fit_transform(df[['City']]).flatten()
```
